# 02 - Character-level tokenizer

**Goal:** turn raw text into integer token IDs (and back), so a network can consume it. This is the tokenizer our tiny GPT will use.

**Key ideas so far**
- A network only does arithmetic, so text must become numbers.
- Production models train a subword tokenizer (e.g. BPE). A character-level tokenizer needs no training: the vocabulary is just the set of distinct characters in the corpus.
- Each token ID is an integer in `[0, vocab_size)`, which is also an index into the model's logits.

**Tasks**
1. Read `data/input.txt` into a string. How many characters long is it?
2. Build the vocabulary: the **sorted** list of distinct characters. Check that `vocab_size == 65`. Print it - which characters surprised you?
3. Build two dicts: `stoi` (char -> int) and `itos` (int -> char).
4. Write `encode(s) -> list[int]` and `decode(ids) -> str`.
5. **Predict first, then check:** what does `encode("hello")` return? (Write the prediction for *which positions hold equal numbers* before running.)
6. Verify the round trip: `decode(encode(text)) == text` for the whole corpus.
7. Encode the whole corpus into a `torch.tensor` of dtype `torch.long`; print its shape and first 20 values.

**Done when:** the round trip holds and you have the full corpus as a tensor of token IDs.

```python
import torch

with open("data/input.txt", encoding="utf-8") as f:
    text = f.read()
```

```python
# task 1
len(text)
```

```python
# task 2
vocab = sorted(set(text))
vocab # there's a 3 in it, surprise
len(vocab) == 65
```

```python
# task 3
stoi = {ch: i for i, ch  in enumerate(vocab)} # char -> int
itos = {i: ch for i,ch in enumerate(vocab)} # int -> char
```

```python
# task 4
def encode(s: str) -> list[int]:
    return [stoi[ch] for ch in s]

def decode(ids: list[int]) -> str:
    return "".join([itos[id] for id in ids])
```
